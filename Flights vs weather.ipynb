```python
# 1. Data ingestion. 
#   A. Dataset downloaded from Kaggle - "Flight Delay Dataset — 2024". File contains over 7 milion rows
file_path = "/Volumes/workspace/default/loty/flight_data_2024.csv"

# Loading the file
df = spark.read.format("csv") \
  .option("header", "true") \
  .option("inferSchema", "true") \
  .option("sep", ",") \
  .load(file_path)

# Previewing data. Medallion Architecture - Brozne layer
display(df)
```

```python
df.createOrReplaceTempView("data")
```

```python
%sql
SELECT year, Count(*) AS number_of_flights 
FROM data GROUP BY year
```

```python
# B. Weather data from open-meteo.com
import requests
import pandas as pd
from pyspark.sql.functions import col, to_timestamp, year, hour

# 1. Airport Configuration (Coordinates)
airports = {
    "JFK": {"lat": 40.6413, "lon": -73.7781},
    "LGA": {"lat": 40.7769, "lon": -73.8740},
    "EWR": {"lat": 40.6895, "lon": -74.1745}
}

START_DATE = "2024-01-01"
END_DATE = "2024-12-31"

all_weather_data = []

# 2. Loop downloading data for each airport
for code, coords in airports.items():
    url = (
        f"https://archive-api.open-meteo.com/v1/archive?"
        f"latitude={coords['lat']}&longitude={coords['lon']}&"
        f"start_date={START_DATE}&end_date={END_DATE}&"
        f"hourly=temperature_2m,precipitation,snow_depth,wind_speed_10m,wind_gusts_10m,visibility&"
        f"timezone=America%2FNew_York"
    )
    
    response = requests.get(url)
    if response.status_code == 200:
        data = response.json()
        pdf = pd.DataFrame(data['hourly'])
        pdf['airport_code'] = code  # We add the airport code to know where the data comes from
        all_weather_data.append(pdf)
    else:
        print(f"Error retrieving data for {code}: {response.status_code}")

# 3. Combining data into one Spark DataFrame
if all_weather_data:
    # We combine Pandas lists into one DF and then to Spark
    combined_pdf = pd.concat(all_weather_data, ignore_index=True)
    df_weather_detailed = spark.createDataFrame(combined_pdf)

    # 4. Time conversion and cleaning
    df_weather_final = df_weather_detailed \
        .withColumn("timestamp", to_timestamp(col("time"), "yyyy-MM-dd'T'HH:mm")) \
        .withColumn("rok", year(col("timestamp"))) \
        .withColumn("godzina", hour(col("timestamp")))

    # 5. Display the results
    display(df_weather_final)
```

```python
# 2. ETL - Silver Layer. Cleaning data
from pyspark.sql.functions import col, count, when, isnan

# 1. I define a list of airports that interest me
ny_airports = ["JFK", "LGA", "EWR"]

# 2. Filtering
df_flights_ny = df.filter(col("Origin").isin(ny_airports))

# 3. Checking how many flights are left after filtering
print(f"Number of flights before filtering: {df.count()}")
print(f"Number of flights from NY airports (after filtering): {df_flights_ny.count()}")
```

```python
# 2B Data Profiling – Data Quality. Missing data analysis
# Checking for missing key columns for lag analysis
from pyspark.sql.functions import col, count, when, isnan

# I select the most important columns for causal analysis
cols_to_check = [
    "dep_delay", "arr_delay", "cancelled", 
    "weather_delay", "nas_delay", "late_aircraft_delay"
]

null_counts = df_flights_ny.select([
    count(when(isnan(c) | col(c).isNull(), c)).alias(c) 
    for c in cols_to_check
])

display(null_counts)
```
